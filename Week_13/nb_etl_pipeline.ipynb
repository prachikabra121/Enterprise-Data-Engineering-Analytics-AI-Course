```python
# from pyspark.sql import functions as F
bronze_order = (
    spark.read.json(sales_path)
    .withColumn("_ingest_ts",F.current_timestamp())
    .withColumn("_source_file",F.expr("_metadata.file_path"))
)

(
bronze_order.write
.format("delta")
.mode("overwrite")
.option("overwriteSchema","true")
.saveAsTable("bronze.sales_orders_raw")
)
```

```python
customers_path = "/databricks-datasets/retail-org/customers/customers.csv"

# sanity check the source file exists before writing
display(dbutils.fs.ls("/databricks-datasets/retail-org/customers/"))

bronze_customers = (
    spark.read.option("header", "true")
    .option("inferSchema", "true")
    .csv(customers_path)
    .withColumn("_ingest_ts", F.current_timestamp())
)

print("rows read:", bronze_customers.count())

(
    bronze_customers.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable("bronze.customers_raw")
)

print(f"bronze.customers_raw rows: {spark.table('bronze.customers_raw').count()}")
```

```python
# ---------------------------------------------------------------------------
# 2. Silver layer - clean, cast, dedupe, flatten nested structures
# ---------------------------------------------------------------------------
spark.sql("""
    CREATE OR REPLACE TABLE silver.sales_orders AS
    SELECT
      order_number,
      customer_id,
      customer_name,
      CAST(FROM_UNIXTIME(TRY_CAST(order_datetime AS BIGINT)) AS TIMESTAMP) AS order_ts,
      CAST(FROM_UNIXTIME(TRY_CAST(order_datetime AS BIGINT)) AS DATE)      AS order_date,
      product.id                             AS product_id,
      product.name                           AS product_name,
      product.curr                           AS currency,
      CAST(product.price AS DECIMAL(10,2))   AS unit_price,
      CAST(product.qty AS INT)               AS quantity,
      CAST(product.price AS DECIMAL(10,2)) * CAST(product.qty AS INT) AS line_revenue
    FROM (
      SELECT *, explode(ordered_products) AS product
      FROM bronze.sales_orders_raw
    )
    WHERE order_number IS NOT NULL
    QUALIFY ROW_NUMBER() OVER (
      PARTITION BY order_number, product.id ORDER BY _ingest_ts DESC
    ) = 1
""")

spark.sql("""
    CREATE OR REPLACE TABLE silver.customers AS
    SELECT
      customer_id,
      customer_name,
      TRIM(state)  AS state,
      TRIM(city)   AS city,
      loyalty_segment
    FROM (
      SELECT *, ROW_NUMBER() OVER (
        PARTITION BY customer_id ORDER BY _ingest_ts DESC
      ) AS rn
      FROM bronze.customers_raw
    )
    WHERE rn = 1 AND customer_id IS NOT NULL
""")

print(f"silver.sales_orders rows: {spark.table('silver.sales_orders').count()}")
print(f"silver.customers rows: {spark.table('silver.customers').count()}")
```

```python
# ---------------------------------------------------------------------------
# 3. Gold layer - business-ready aggregates
# ---------------------------------------------------------------------------
spark.sql("""
    CREATE OR REPLACE TABLE gold.daily_sales_by_state AS
    SELECT
      o.order_date,
      c.state,
      COUNT(DISTINCT o.order_number) AS num_orders,
      SUM(o.quantity)                AS units_sold,
      ROUND(SUM(o.line_revenue), 2)  AS total_revenue
    FROM silver.sales_orders o
    JOIN silver.customers c USING (customer_id)
    GROUP BY o.order_date, c.state
""")

spark.sql("""
    CREATE OR REPLACE TABLE gold.top_products AS
    SELECT
      product_id,
      product_name,
      SUM(quantity)               AS units_sold,
      ROUND(SUM(line_revenue), 2) AS total_revenue
    FROM silver.sales_orders
    GROUP BY product_id, product_name
    ORDER BY total_revenue DESC
""")

```

```python
# ---------------------------------------------------------------------------
# 4. Validate
# ---------------------------------------------------------------------------
for table in [
    "bronze.sales_orders_raw", "bronze.customers_raw",
    "silver.sales_orders", "silver.customers",
    "gold.daily_sales_by_state", "gold.top_products",
]:
    print(f"{table:<30} {spark.table(table).count():>8} rows")

spark.table("gold.top_products").show(10, truncate=False)
```
