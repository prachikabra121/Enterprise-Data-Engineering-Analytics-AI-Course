<h1>3. Basic Concepts</h1>
Before jumping into plots, let’s break down the structure of a typical Matplotlib visualization:

1)Figure: The overall window or page where your plot appears.

2)Axes: The area where the data is plotted. A figure can have multiple axes (subplots).

3)Elements: Titles, labels, legends, gridlines, etc.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="darkgrid")
```

```python
#Let’s start with a simple line plot:
```

```python
import matplotlib.pyplot as plt

# Data to plot
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# Create a plot
plt.plot(x, y)  # x is the horizontal axis, y is the vertical axis

# Add labels and title
plt.title('My First Plot')  # Title of the plot
plt.xlabel('X-axis')  # Label for the horizontal axis
plt.ylabel('Y-axis')  # Label for the vertical axis

# Show the plot
plt.show()
```

Explanation:

-plt.plot() creates the line graph.

-plt.title(), plt.xlabel(), and plt.ylabel() add helpful text.

-plt.show() displays the plot.


<h1>5. Types of Plots</h1>

1)Line Plot

Used to show trends or changes over time.

```python
plt.plot(x, y, color='green', marker='o', linestyle='--')
plt.title('Line Plot Example')
plt.show()
```

```python
plt.plot(x, y, color='green', marker='x', linestyle='-')
plt.title('Line Plot Example')
plt.show()
```

2)Scatter Plot

Shows individual points to highlight patterns or relationships.

```python
plt.scatter(x, y, color='red', label='Data Points')
plt.title('Scatter Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend()  # Add a legend
plt.show()
```

3)Bar Plot

Useful for comparing categories.

```python
categories = ['A', 'B', 'C', 'D']
values = [10, 20, 15, 25]

plt.bar(categories, values, color='blue')
plt.title('Bar Chart Example')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.show()
```

4)Histogram

Displays data distribution.

```python
data = [1, 2, 2, 3, 3, 3, 4, 4, 5]

plt.hist(data, bins=5, color='purple', edgecolor='black')
plt.title('Histogram')
plt.show()
```

6. Customizing Plots
   
Matplotlib gives you full control over the look and feel of your plots.

Adding Gridlines

```python
plt.plot(x, y)
plt.grid(True)  # Enable gridlines
plt.show()
```

```python
plt.plot(x, y, label='Line 1', color='blue')
plt.plot(x, [1, 3, 5, 7, 9], label='Line 2', color='red')
plt.legend(loc='lower right')  # Add a legend in the top-left corner
plt.show()
```

7. Multiple Plots (Subplots)

You can organize multiple plots within the same figure:

```python
plt.subplot(1, 2, 1)  # 1 row, 2 columns, this is the 1st plot
plt.plot(x, y)
plt.title('Plot 1')

plt.subplot(1, 2, 2)  # 1 row, 2 columns, this is the 2nd plot
plt.bar(categories, values)
plt.title('Plot 2')

plt.tight_layout()  # Adjust spacing between plots
plt.show()
```

```python
#Control the placement and appearance of ticks:
plt.plot([1, 2, 3, 4], [10, 20, 25, 30])

# Customize tick positions and labels
plt.xticks([1, 2, 3, 4], ['One', 'Two', 'Three', 'Four'], rotation=45)  # Rotate labels
plt.yticks([10, 20, 30], ['Low', 'Medium', 'High'])

plt.title('Custom Tick Labels')
plt.show()
```

```python
#Spines are the lines defining the axis boundaries:
fig, ax = plt.subplots()

ax.plot([1, 2, 3, 4], [10, 20, 25, 30])

# Remove top and right spines
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Change spine position
ax.spines['left'].set_position(('outward', 10))
ax.spines['bottom'].set_position(('outward', 10))

plt.title('Custom Spines')
plt.show()

```

```python
#Matplotlib supports interactive features like zooming, panning, and event handling.
import 
#Interactive Mode
#Enable interactive plotting:
plt.ion()  # Interactive mode on
plt.plot([1, 2, 3], [4, 5, 6])
plt.title('Interactive Mode')
plt.show()
```

```python
def on_click(event):
    print(f"Mouse clicked at {event.xdata}, {event.ydata}")

fig, ax = plt.subplots()
ax.plot([1, 2, 3], [4, 5, 6])

# Connect the click event to the function
fig.canvas.mpl_connect('button_press_event', on_click)
plt.show()
```

```python
fig, ax1 = plt.subplots()

# First plot
ax1.plot([1, 2, 3], [4, 5, 6], color='blue')
ax1.set_ylabel('Blue Axis', color='blue')

# Second y-axis
ax2 = ax1.twinx()
ax2.plot([1, 2, 3], [10, 20, 30], color='red')
ax2.set_ylabel('Red Axis', color='red')

plt.title('Twin Axes Example')
plt.show()
```

```python
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

x = [1, 2, 3, 4]
y = [4, 5, 6, 7]
z = [7, 8, 9, 10]

ax.plot(x, y, z)
plt.title('3D Line Plot')
plt.show()
```

```python
import numpy as np

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Generate data
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))

# Plot surface
ax.plot_surface(X, Y, Z, cmap='viridis')
plt.title('3D Surface Plot')
plt.show()
```

```python
import matplotlib.gridspec as gridspec

fig = plt.figure()

gs = gridspec.GridSpec(3, 3)
ax1 = fig.add_subplot(gs[0, :])  # First row
ax2 = fig.add_subplot(gs[1, :-1])  # Second row, first two columns
ax3 = fig.add_subplot(gs[1:, -1])  # Second and third rows, last column
ax4 = fig.add_subplot(gs[-1, 0])  # Last row, first column
ax5 = fig.add_subplot(gs[-1, -2])  # Last row, second column

plt.tight_layout()
plt.show()
```
