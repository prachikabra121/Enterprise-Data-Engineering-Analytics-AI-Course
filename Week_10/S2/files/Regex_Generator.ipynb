<h1>Regex in Python</h1>

Regex (Regular Expressions) in Python are powerful tools used to match patterns in strings. They allow searching, extracting, replacing, or validating text in an efficient and flexible manner.

Python provides the re module for working with regex.

<h1>Basic Regex Functions</h1>
Here are the key functions in the re module:

<h4>re.search(pattern, string)</h4>

Searches for the first occurrence of the pattern in the string.
Returns a match object if found, otherwise returns None.
<h4>re.match(pattern, string)</h4>

Checks if the pattern matches from the beginning of the string.
Returns a match object if found, otherwise returns None.
<h4>re.findall(pattern, string)</h4>

Returns a list of all occurrences of the pattern in the string.
<h4>re.finditer(pattern, string)</h4>

Returns an iterator yielding match objects for all matches of the pattern.
<h4>re.sub(pattern, replacement, string)</h4>

Replaces occurrences of the pattern with the replacement string.
<h4>re.split(pattern, string)</h4>

Splits the string wherever the pattern matches.
<h4>re.compile(pattern)</h4>

Compiles the regex pattern into a regex object for reuse.

```python
#1. Searching for a pattern
import re

text = "The train in Spain train"
match = re.search(r"rain", text)
if match:
    print("Found:", match.group()) # Output: Found: rain
```

```python
#1. Searching for a pattern
import re

text = "The rain in Spain rain"
matches = re.findall(r"rain", text)
if len(matches)>=2:
    print("Found:", matches[1])  # Output: Found: rain
```

```python
import re

text = "The rain in Spain rain"

match = re.match(r"The", text)

if match:
    print("Matched:", match.group())
```

```python
text = "The rain in spain stays mainly in the plain"
pattern = r"pain"

for match in re.finditer(pattern,text):
    print("Match: ",match.group(),"| start :",match.start())
```

```python
text = 'My 123 number is 12345 and yours is 45678'
patten = r"(\d+)"

for match in re.finditer(patten,text):
    print("Match: ",match.group())
```

```python
#Simple email validation using regex (note: this is basic, real-world email regex can be more complex):
import re

email = "9pr900achi900@chetu.com"
pattern = r"^[a-zA-Z][a-zA-Z0-9_.+-]+@(gmail|yahoo|chetu)\.com$"

if re.match(pattern, email):
    print("Valid email!")
else:
    print("Invalid email!")
```

[a-zA-Z0-9_.+-]+:

[a-zA-Z0-9_.+-]: This part defines a character class that matches any of the following characters:
Lowercase letters (a-z)
Uppercase letters (A-Z)
Digits (0-9)
Underscores (_)
Periods (.)
Plus signs (+)
Hyphens (-)
+: The + means one or more of the preceding characters. So this part of the pattern matches the local part of the email (the part before the @ symbol), which can consist of one or more characters from the defined set.
@:

This matches the literal @ symbol that separates the local part and the domain part of the email.
[a-zA-Z0-9-]+:

This part matches the domain name (after the @ symbol), which can consist of:
Lowercase letters (a-z)
Uppercase letters (A-Z)
Digits (0-9)
Hyphens (-)
The + ensures that there is at least one character in the domain name.
\.:

This part matches the literal period (.) that separates the domain name from the top-level domain (like .com, .org, etc.). The backslash \ is used to escape the period since the period is a special character in regex, meaning "any character".
[a-zA-Z0-9-.]+:

This part matches the top-level domain (TLD), such as .com or .org. It can consist of:
Lowercase letters (a-z)
Uppercase letters (A-Z)
Digits (0-9)
Periods (.)
Hyphens (-)
The + ensures that the TLD consists of one or more of these characters.
if re.match(pattern, email)::

re.match() checks whether the beginning of the string (email) matches the specified regex pattern (pattern).
If there is a match, it returns a match object. If there is no match, it returns None.
In this case, since "user@example.com" follows the pattern of a typical email address, re.match() will return a match.
print("Valid email!"):

If a match is found (i.e., the email address matches the pattern), it prints "Valid email!".
else: print("Invalid email!"):

If no match is found (i.e., the email address does not follow the pattern), it prints "Invalid email!".

```python
🔹 What is a Generator?

A generator in Python is a special type of function that lets you iterate (loop) over data one item at a time, without storing everything in memory at once.

They are created using the yield keyword.

They are memory efficient because they generate values on the fly (lazy evaluation).

Generators are iterators, but not all iterators are generators.
```

```python
---1)large data set not efficient storing evreything in memory 
---1) generator produce values only when needed,saving memory and improve performance
```

```python
def square_list(n):
    result=[]
    for i in range(n):
        result.append(i*i)
    return result
```

```python
square_list(10)
```

```python
def square_gen(n):
   for i in range(n):
       yield i*i
```

```python
gen = square_gen(10)
```

```python
for value in gen:
    print(value,end=" ")
```

```python
next(gen)
```

```python
def infinite_number(start=1):
    while True:
        yield start
        start +=1
    
```

```python
gen = infinite_number()
```

```python
for i in range(5):
    print(next(gen))
```

```python
import csv

def read_csv_gen(file_path):
    with open(file_path,mode='r') as file:
        reader = csv.reader(file)
        for row in reader:
            yield row
```

```python
gen = read_csv_gen("goodwill_05-04-25_out.csv")
```

```python
next(gen)
```

```python
for i,row in enumerate(gen,start=1):
    if i == 5:
        print(row)
        break
    
    
```
