```python
import pandas as pd
import numpy as np

data = {
    'Employee_ID': [101, 102, 103, 104, 105],
    'Employee_Name': ['Anand', 'Rahul', None, 'John', 'Priya'],
    'Department': ['IT', 'HR', 'Finance', None, 'IT'],
    'Salary': [80000, np.nan, 90000, 75000, np.nan]
}
```

```python
df2 = pd.DataFrame(data)
```

```python
df2
```

```python
df2.isnull().sum()
```

```python
SELECT
SUM(CASE WHEN Employee_Name IS NULL THEN 1 ELSE 0 END) AS Missing_Name,
SUM(CASE WHEN Salary IS NULL THEN 1 ELSE 0 END) AS Missing_Salary
FROM Employee;
```

```python
cleaned_df=df2.dropna(subset=['Employee_Name'])
```
